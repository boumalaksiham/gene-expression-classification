# Gene Expression Classification — Golub Leukemia Dataset

This notebook accompanies the reproducible command-line analysis in `scripts/run_analysis.py`.

The task is to distinguish **ALL** from **AML** using high-dimensional microarray gene-expression profiles.

## Why this is a difficult ML problem

The cohort contains only 72 patients but 7,129 expression features. This is a classic `p >> n` setting, so model evaluation and leakage-safe feature selection are more important than simply fitting a high-capacity classifier.

```python
from pathlib import Path
import subprocess, sys
ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
subprocess.run([sys.executable, str(ROOT / 'scripts' / 'run_analysis.py')], check=True)
```
